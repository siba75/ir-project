# Quora Full Evaluation

Dataset: `beir/quora/test`

Official dataset page: https://ir-datasets.com/beir.html#beir/quora/test

This notebook is prepared for the interview requirement: evaluation must use all queries available in the qrels file.

```python
from pathlib import Path
import json

BASE_DIR = Path(r"D:\\ir-project")
qrels_path = BASE_DIR / "datasets" / "quora" / "qrels.json"

with open(qrels_path, "r", encoding="utf-8") as file:
    qrels = json.load(file)

query_ids = list(qrels.keys())
total_qrel_judgments = sum(len(doc_ids) for doc_ids in qrels.values())

print("Dataset: beir/quora/test")
print(f"Local qrels file: {qrels_path}")
print(f"Total qrels queries: {len(qrels)}")
print(f"Total qrel judgments: {total_qrel_judgments}")
print(f"Queries selected for final evaluation: {len(query_ids)}")
print("Evaluation scope: all_qrels_queries")

assert len(query_ids) == len(qrels)
assert len(query_ids) == 10000
```

## Start Full Evaluation

Run the next cell to start the full evaluation in a separate process. This avoids crashing the Jupyter kernel and writes progress to a log file.

If the evaluation stops, run the same cell again. The script resumes from saved checkpoints in `reports/evaluation_checkpoints`.

```python
from pathlib import Path
import subprocess

BASE_DIR = Path(r"D:\\ir-project")
python_exe = Path(r"D:\\anaconda3\\envs\\ir-project\\python.exe")
script_path = BASE_DIR / "services" / "evaluation_service" / "evaluate_dataset_direct.py"
log_path = BASE_DIR / "reports" / "quora_full_evaluation.log"
report_path = BASE_DIR / "reports" / "quora_evaluation_results.json"

cmd = [
    str(python_exe),
    str(script_path),
    "quora",
    "--progress-every",
    "100",
    "--checkpoint-dir",
    str(BASE_DIR / "reports" / "evaluation_checkpoints"),
]

log_path.parent.mkdir(parents=True, exist_ok=True)
log_file = open(log_path, "w", encoding="utf-8")
process = subprocess.Popen(
    cmd,
    cwd=str(BASE_DIR),
    stdout=log_file,
    stderr=subprocess.STDOUT,
)
log_file.close()

print("Full evaluation started successfully.")
print(f"PID: {process.pid}")
print("Dataset: beir/quora/test")
print("Queries selected for final evaluation: 10000")
print("Evaluation scope: all_qrels_queries")
print(f"Progress log: {log_path}")
print(f"Checkpoint directory: {BASE_DIR / 'reports' / 'evaluation_checkpoints'}")
print(f"Final report will be saved to: {report_path}")
```

Run this cell any time while the evaluation is running to show the latest progress lines.
